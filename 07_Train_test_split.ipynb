# Split test train by country

```python
def split_within_country_and_fix_clusters(
    df, country_col, cluster_col, train_frac=0.7, random_state=None
):
    if random_state is not None:
        np.random.seed(random_state)

    train_idx = []
    test_idx = []

    # Step 1: random 70/30 split inside each country
    for country, df_country in df.groupby(country_col):
        train_rows = df_country.sample(frac=train_frac, random_state=random_state).index
        test_rows = df_country.drop(train_rows).index

        train_idx.extend(train_rows)
        test_idx.extend(test_rows)

    train_idx = set(train_idx)
    test_idx = set(test_idx)

    # Step 2: ensure all clusters appear in both sets
    train_clusters = set(df.loc[list(train_idx), cluster_col])
    test_clusters = set(df.loc[list(test_idx), cluster_col])
    all_clusters = set(df[cluster_col].unique())

    # Clusters missing in train
    missing_train = all_clusters - train_clusters
    # Clusters missing in test
    missing_test = all_clusters - test_clusters

    # Step 3: fix missing clusters by moving a few samples
    for cl in missing_train:
        # move one sample of this cluster from test to train
        candidate = df[(df.index.isin(test_idx)) & (df[cluster_col] == cl)].index
        if len(candidate) > 0:
            move = candidate[0]
            test_idx.remove(move)
            train_idx.add(move)

    for cl in missing_test:
        # move one sample of this cluster from train to test
        candidate = df[(df.index.isin(train_idx)) & (df[cluster_col] == cl)].index
        if len(candidate) > 0:
            move = candidate[0]
            train_idx.remove(move)
            test_idx.add(move)

    # Final boolean masks
    train_mask = df.index.isin(train_idx)
    test_mask = df.index.isin(test_idx)

    return train_mask, test_mask
```

```python
# 1. Split by country
train_mask, test_mask = split_within_country_and_fix_clusters(
    df=all_data,
    country_col="country_iso3",
    cluster_col="cluster4",
    train_frac=0.7,
    random_state=None   # leave None for a different split every run
)

X_train = all_data.loc[train_mask, predictors]
y_train = all_data.loc[train_mask, ['UMAP1','UMAP2','UMAP3']]

X_test  = all_data.loc[test_mask, predictors]
y_test  = all_data.loc[test_mask, ['UMAP1','UMAP2','UMAP3']]
```
