```python
# Import the area of interest
aoi= ee.Image('projects/ee-benitezl/assets/Floristics/review/aoi')

# And the aoi masked to remove low biomass areas/ tropical wet forest
landscape = ee.Image('projects/ee-benitezl/assets/Floristics/review/masked_aoi')
# These are created in this earth engine script: https://code.earthengine.google.com/fe88b4d8652fd46291e52d5c2399a09d
```

# Maps for UMAP 1, 2, and 3

```python
# Load in three layers
UMAP1 = ee.Image('projects/ee-benitezl/assets/Floristics/review/UMAP1')
UMAP2 = ee.Image('projects/ee-benitezl/assets/Floristics/review/UMAP2')
UMAP3 = ee.Image('projects/ee-benitezl/assets/Floristics/review/UMAP3')
```

```python
# Mask by aoi
UMAP1= UMAP1.updateMask(aoi.eq(1))
UMAP2= UMAP2.updateMask(aoi.eq(1))
UMAP3= UMAP3.updateMask(aoi.eq(1))
```

```python
# Create an interactive map to see three layers
Map = geemap.Map()

# Visualization parameters for each band
vis_red   = { 'palette': ['black', 'red']}
vis_green = { 'palette': ['black', 'green']}
vis_blue  = { 'palette': ['black', 'blue']}

# Add the image with color scales
Map.addLayer(UMAP1, vis_red,   'UMAP1 (red)')
Map.addLayer(UMAP2, vis_green, 'UMAP2 (green)')
Map.addLayer(UMAP3, vis_blue,  'UMAP3 (blue)')

# Center the map
Map.centerObject(UMAP1, 7)

Map

```

```python
# Make rgb by combining three bands
rgb = UMAP1.addBands(UMAP2).addBands(UMAP3)
```

```python
# Create an interactive map for RGB image
Map = geemap.Map()

Map.add_basemap("SATELLITE")

# Visualization parameters for each band
vis_red   = { 'palette': ['yellow']}
vis_green = { 'palette': [ 'green']}

# RGB visualization parameters
rgb_vis = { 'bands': ['UMAP1', 'UMAP2', 'UMAP3'],'min': [-3.5,-7.1,-2.1], 'max': [2.03,4.38,3.04], 'gamma': [1, 1, 1]}

# Add the RGB composite
Map.addLayer(rgb, rgb_vis, 'UMAP RGB')

# Center the map
Map.centerObject(UMAP1, 7)

Map
```

```python
# Mask RGB to remove agriculture and other low biomass areas, plus tropical wet forests
rgb_landscape= rgb.updateMask(landscape.eq(1))
```

```python
# Create an interactive map
Map = geemap.Map()

Map.add_basemap("SATELLITE")


# RGB visualization parameters
rgb_vis = { 'bands': ['UMAP1', 'UMAP2', 'UMAP3'],'min': [-3.5,-7.1,-2.1], 'max': [2.03,4.38,3.04], 'gamma': [1, 1, 1]}

# Add the RGB composite
Map.addLayer(rgb_landscape, rgb_vis, 'UMAP RGB')

# Center the map
Map.centerObject(UMAP1, 7)

Map
```
